```python
customers_data = [
    ("C00001", "Aman Retail", "Hyderabad", "Corporate"),
    ("C00002", "Blue Mart", "Mumbai", "Retail"),
    ("C00003", "Fresh Basket", "Bangalore", "Retail"),
    ("C00004", "Prime Stores", "Delhi", "SMB"),
    ("C00005", "Quick Shop", "Pune", "Corporate")
]

customers_columns = [
    "customer_id",
    "customer_name_master",
    "customer_city",
    "segment"
]

customers_df = spark.createDataFrame(
    customers_data,
    customers_columns
)

customers_df.show()
```

```python
joined_df=df.join(
    customers_df,
    on="customer_id",
    how="inner"
)
joined_df.show()
```

```python
left_join_df = df.join(
    customers_df,
    on="customer_id",
    how="left"
)

left_join_df.show()
```

```python
right_join_df = df.join(
    customers_df,
    on="customer_id",
    how="right"
)

right_join_df.show()
```

```python
full_join_df=df.join(
    customers_df,
    on="customer_id",
    how="outer"
)
full_join_df.show()
```

```python
from pyspark.sql.functions import col, to_date

df2 = df.withColumn(
    "order_date",
    to_date(col("order_date"), "yyyy-MM-dd")
)
```

```python
from pyspark.sql.functions import year

df2.withColumn(
    "year",
    year("order_date")
).select(
    "order_date",
    "year"
).show()
```

```python
from pyspark.sql.functions import month

df2.withColumn(
    "month",
    month("order_date")
).select(
    "order_date",
    "month"
).show()
```

```python
from pyspark.sql.functions import dayofmonth

df2.withColumn(
    "day",
    dayofmonth("order_date")
).select(
    "order_date",
    "day"
).show()
```
