```python
from pyspark.sql import SparkSession
spark=SparkSession.builder\
      .appName("python training")\
      .getOrCreate()
```

```python
print(spark)
print(spark.version)
```

```python
data = [
    (101, "Laptop", "Electronics", 65000),
    (102, "Mouse", "Accessories", 1500),
    (103, "Monitor", "Electronics", 18000),
    (104, "Office chair", "Furniture", 9000),
    (105, "Keyboard", "Accessories", 2500)
]

columns = [
    "product_id",
    "product_name",
    "category",
    "price"
]
```

```python
df = spark.createDataFrame(data, columns)
```

```python
df.show()
```

```python
df.select("product_id","product_name").show()
```

```python
df.filter(df.product_id==101).show()
```

```python
df.filter((df.product_id >100) &(df.price>2000)).show()
```

```python
from pyspark.sql.types import StructType, StructField, IntegerType, StringType,DoubleType, BooleanType
```

```python
schema=StructType([
    StructField("Product_id",IntegerType(),False),
    StructField("ProductName",StringType(),False),
    StructField("Category",StringType(),True),
    StructField("Price",DoubleType(),True),
    StructField("Availability",BooleanType(),True)

])
```

```python
from pyspark.sql.functions import col

df=df.withColumn(
    "Price_with_column",
    col("Price")*1.38

)
df.show()
```

```python
df1=df.withColumn(
    "Price",
    col("Price")*0.28
)
df1.show()
```
