```python
df2=spark.read.csv(
    "retail_transactions_100k.csv",
    header=True,
    inferSchema=True
)
df2.show()
```

```python
#1
df2.show(20)
#2
df2.printSchema()
#3
print(df2.columns)
#4
print(df2.count())
#5
df2.select("order_id", "product_name", "city", "quantity", "unit_price").show()
#6
df2.filter(df2.category=="Electronics").show()
#7
df2.filter(df2.city=="Hyderabad").show()
#8
df2.filter(df2.quantity>3).show()
#9
df2.filter(df2.unit_price>20000).show()
#10
df2.filter(df2.status=="Delivered").show()
```

```python
#set 2
#11
df2.filter((df2.category=="Electronics")&(df2.city=="Hyderabad")).show()
#12
df2.filter((df2.region=="South")&(df2.status=="Delivered")).show()
#13
df2.filter((df2.quantity>2)&(df2.unit_price>10000)).show()
#14
df2.filter((df2.city==" Mumbai")|(df2.city=="Pune")).show()
#15
df2.filter(df2.status!="Cancelled").show()
#16
df2.filter((df2.category=="Furniture")&(df2.unit_price>8000)).show()
#17
df2.filter((df2.category=="Accessories")&(df2.quantity>3)).show()
```

```python
#set 3
#18
df2.select("order_id","customer_name","product_name","unit_price").show()
#19
df2.withColumnRenamed("unit_price","price").show()
#20
df2.withColumnRenamed("customer_name","customer").show()
#21
df2.drop("rating").show()
#22
df_dropped = df2.drop("customer_name", "rating", "payment_method")
df_dropped.show()
```

```python
#set 4
# 23
df_calc = df2.withColumn("gross_amount", df2.quantity * df2.unit_price)
df_calc.show()
# 24
df_calc = df_calc.withColumn("discount_amount", df_calc.gross_amount * df_calc.discount_pct / 100)
df_calc.show()
# 25
df_calc = df_calc.withColumn("final_amount", df_calc.gross_amount - df_calc.discount_amount)
df_calc.show()
# 26
from pyspark.sql.functions import lit

df_calc = df_calc.withColumn("country", lit("India"))
df_calc.show()

# 27
from pyspark.sql.functions import when, col

df_calc = df_calc.withColumn(
    "high_value",
    when(col("final_amount") >= 50000, "Yes").otherwise("No")
)
df_calc.show()
```

```python
#28
df_calc = df_calc.withColumn(
    "order_size",
    when(col("final_amount") >= 50000, "High")
    .when(col("final_amount") >= 15000, "Medium")
    .otherwise("Low")
)
df_calc.select("order_id", "final_amount", "order_size").show()
#29
df_calc = df_calc.withColumn("quantity_type", when(col("quantity") >= 4, "Bulk").otherwise("Regular"))
df_calc.select("order_id", "quantity", "quantity_type").show()
#30
df_calc = df_calc.withColumn(
    "delivery_category",
    when(col("status") == "Delivered", "Completed")
    .when(col("status").isin("Cancelled", "Returned"), "Failed")
    .otherwise("In Progress")
)
df_calc.select("order_id", "status", "delivery_category").show()
```
